```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

df = df.rename(columns={
    'purchase_amount_(usd)': 'purchase_amount'
})

print(df.columns.tolist())
```

```python
df.columns=df.columns.str.lower();
df.columns=df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_usd': 'purchase_amount'})
```

```python
df.columns
```

```python
# create new column  for age group
labels=['Young Adult','Adult','Middle aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df.head()
```

```python
# create column purchace_frequency_days

frequency_mapping = {
    'Daily': 1,
    'Weekly': 7,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Annually': 365
}

df['purchace_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
